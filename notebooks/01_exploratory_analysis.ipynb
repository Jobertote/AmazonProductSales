# Exploración y validación del dataset de ventas

Este notebook documenta la exploración inicial del archivo `clean_dataset.csv`, generado durante el proceso de limpieza. Se valida la estructura disponible y se calculan indicadores descriptivos sobre promociones, productos destacados, patrocinio, precios, compras y reseñas.

## Alcance de la exploración

El análisis se organiza en tres pasos: carga del conjunto de datos y revisión de sus columnas; resumen de la presencia de cupones, productos *best seller* y anuncios patrocinados; y cálculo de valores extremos y acumulados para las variables numéricas principales. Los resultados permiten comprobar, a alto nivel, el contenido del dataset antes de utilizarlo en análisis posteriores.

```python
from amazon_product_sales.utils.get_paths import build_paths
import pandas as pd

paths = build_paths()

df_cleaned = pd.read_csv(paths['processed'] / "clean_dataset.csv")

df_cleaned.columns
```

## Indicadores de visibilidad comercial

A continuación se resume el total de productos y la cantidad —junto con su porcentaje sobre el total— de registros que tienen cupón, la etiqueta *best seller* o patrocinio. Las columnas `has_coupon`, `best_seller` y `sponsored` se tratan como indicadores booleanos, por lo que su suma equivale al número de productos que cumplen cada condición.

```python
total_recors = len(df_cleaned)

main_stats = pd.DataFrame({
    "total_records": [len(df_cleaned)],
    "coupon_records": [(df_cleaned["has_coupon"]).sum()],
    "coupon_porcentage": [round((df_cleaned["has_coupon"]).sum() / total_recors * 100,2)],
    "best_seller_records": [(df_cleaned["best_seller"]).sum()],
    "best_seller_porcentage": [round((df_cleaned["best_seller"]).sum() / total_recors * 100,2)],
    "sponsor_records": [(df_cleaned["sponsored"]).sum()],
    "sponsor_porcentage": [round((df_cleaned["sponsored"]).sum() / total_recors * 100,2)],
})

main_stats
```

## Rango y volumen de actividad

Esta sección calcula los valores máximo y mínimo de `current_discounted_price_f` y de `bought_last_month`, además de las compras y reseñas acumuladas. Estos indicadores ayudan a identificar la escala del catálogo y posibles valores extremos que convendría revisar en una exploración más detallada.

```python
max_discount = df_cleaned["current_discounted_price_f"].max()
min_discount = df_cleaned["current_discounted_price_f"].min()
max_bought = df_cleaned["bought_last_month"].max()
min_bought = df_cleaned["bought_last_month"].min()
total_bought = df_cleaned["bought_last_month"].sum()
total_reviews = df_cleaned["number_reviews"].sum()

extra_stats = pd.DataFrame ({
    "max_discount": [max_discount],
    "min_discount": [min_discount],
    "max_bought": [max_bought],
    "min_bought": [min_bought],
    "total_bought": [total_bought],
    "total_reviews": [total_reviews],
})

extra_stats
```
